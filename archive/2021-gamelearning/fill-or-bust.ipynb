```python
N = 1000
anna_wins = 0
john = DumbAgent("John")
anna = SmartAgent("Anna", alpha=0.2, epsilon=0.2)
agents = [john, anna]
for n in range(N):
    if (100 * n / N) % 10 == 0:
        print(f"{100 * n / N:.1f}%, ", end="")
    game = FillOrBust(agents, max_score=10000)
    while not game.game_is_over:
        agent = game.current_player
        action = agent.select_action(game.state)
        game.perform_action(action)

    if anna.score >= max(player.score for player in game.players):
        anna_wins += 1

print(f"\nAnna win percentage: {100 * anna_wins / n: 0.1f}%")
```
