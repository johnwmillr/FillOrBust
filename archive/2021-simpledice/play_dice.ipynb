```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

from dice.game import play, play_many_games
```

# Make two strategies compete

```python
s1 = play_many_games(0)
s2 = play_many_games(100)
games = (
    pd.DataFrame({"s1": s1, "s2": s2})
    .assign(
        s1_wins=lambda df: df["s1"] > df["s2"]
    )
)
games["s1_wins"].value_counts(True).sort_index()
```

# Average score as a function of stopping score

```python
strats = range(0, 1000, 50)
scores = [play_many_games(strat) for strat in strats]
results = (
    pd.DataFrame({
        "strategy": strats,
        "score": scores})
    .explode("score")
    .assign(score=lambda df: df["score"].astype(int)))
stats = results.groupby("strategy").agg(["mean", "max", "std"])["score"].reset_index()
```

```python
y = "mean"
fig, ax = plt.subplots(figsize=[12, 6])
ax.errorbar(stats["strategy"], stats[y], yerr=stats["std"])
ax.set_ylabel(y, fontsize=16)
ax.set_ylim([0, ax.get_ylim()[-1]])
ax.legend([])
ax.grid(True, alpha=0.6)
plt.show()
```
