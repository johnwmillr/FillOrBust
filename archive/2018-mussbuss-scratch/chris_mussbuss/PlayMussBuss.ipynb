```python
def getDice():    
    # Get the dice values
    dice = browser.find_elements_by_tag_name("img")
    vals = [int(d.get_property("alt").split('-',1)[-1]) for d in dice]    
    return vals, dice

def dieIsHeld(dice_im):    
    regex = re.compile('margin: (\d{2})px')
    return regex.findall(dice_im.get_attribute('style')) == []

def holdDice():
    # Get the hold button for each die
    dice_buttons = divs[1:7]
    
    # Hold all pointers
    for db,img,val in zip(dice_buttons, dice_imgs, dice_values):
        print("\nValue: {v}".format(v=val))
#         if val in [1,5] and not dieIsHeld(img):
        if not dieIsHeld(img):            
            print(" (Holding.)")
            db.click()
            time.sleep(1)
            
def numRemaining():
    return sum([not dieIsHeld(d) for d in dice_imgs])
```

```python
class Player(object):
    
    def __init__(self, name):
        self.name = name
        self._score = 0
        
    def takeTurn(self):
        num_dice = numRemaining()
        print("{n} dice remaining. ".format(n=num_dice), end='')
        if num_dice > 2:
            print("Rolling.")
        else:
            print("Stopping.")
        
    @property
    def score(self):
        return self._score
        
    def __repr__(self):
        return "({n}) Score: {s}".format(n=self.name, s=self.score)
    
    def __str__(self):
        return self.__repr__()    
```
