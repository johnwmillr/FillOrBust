```python
from selenium import webdriver
```

```python
import time
import re
```

```python
browser = webdriver.Chrome()
url = 'https://muss-buss.herokuapp.com/games/new'
browser.get(url)
```

```python
body = browser.find_element_by_tag_name('body')
```

```python
labels = body.find_elements_by_tag_name('label')
```

```python
start = labels[-1].find_element_by_tag_name("input")
```

```python
# Start the game (Loads new URL)
start.click()
time.sleep(1)
start.click()
print(browser.current_url)
```

```python
# Game controls
game = browser.find_element_by_id("game")
```

```python
# Get the ROLL and STOP buttons
divs = game.find_elements_by_tag_name("div")
roll = [d for d in divs if d.text=="ROLL"][0]
stop = [d for d in divs if d.text=="STOP"][0]    
```

```python
def dieIsHeld(dice_im):    
    regex = re.compile('margin: (\d{2})px')
    return regex.findall(dice_im.get_attribute('style')) == []
            
def numRemaining(imgs):
    return sum([not dieIsHeld(d) for d in imgs])
```

```python
class Player(object):
    
    def __init__(self, name):
        self.name = name
        self._score = 0
        self.dice_values = None
        self.dice_imgs = None
        self._stopped = False
                    
    def holdDice(self):
        # Get the hold button for each die
        dice_buttons = divs[1:7]

        # Hold all pointers
        for db,img,val in zip(dice_buttons, self.dice_imgs, self.dice_values):
            print("\nValue: {v}".format(v=val), end='')
            if not dieIsHeld(img):            
                print(" (Holding.)")
                db.click()
                time.sleep(1)
            else:
                print(' ')                
                
    def setDice(self):    
        # Get the dice values
        self.dice_imgs = browser.find_elements_by_tag_name("img")
        self.dice_values = [int(d.get_property("alt").split('-',1)[-1]) for d in self.dice_imgs]
    
    def roll(self):
        roll.click()
        time.sleep(1)
        self.setDice()
        
    def stop(self):
        stop.click()
        time.sleep(1)
                                                  
    def takeTurn(self):
        self.setDice()
        num_dice = numRemaining(self.dice_imgs) if self.dice_imgs is not None else 6
        print("{n} dice remaining. ".format(n=num_dice), end='')
        self.holdDice()
        if num_dice > 2:
            print("Rolling.")
            self.roll()
        else:
            print("Stopping.")
            self.stop()
            
    @property
    def isStopped(self):
        return self._stopped
                                    
    @property
    def score(self):
        return self._score
        
    def __repr__(self):
        return "({n}) Score: {s}".format(n=self.name, s=self.score)
    
    def __str__(self):
        return self.__repr__()    
```

```python
p2 = Player("Chris")
```

```python
p2.takeTurn()
```
