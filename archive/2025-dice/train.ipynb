```python
from dice.agents import Agent, SmartAgent
from dice.train import train, plot_training_metric
from dice.types import DiceCounts, Observation
```

# Train an agent

```python
n_episodes = 1_000
dumb = Agent("dumb")
train(dumb, n_episodes=n_episodes)
plot_training_metric(dumb, "turn_count", downsample=True)
```

```python
n_episodes = 3_000
smart = SmartAgent("smart", learning_rate=0.01, epsilon=0.6, discount_factor=0.9)
train(smart, n_episodes=n_episodes)
plot_training_metric(smart, "turn_count", downsample=True)
```

# Inspect the q-table

```python
obs = Observation(DiceCounts(), held_score=50, last_action="hold")
smart.q_values[obs]
```
